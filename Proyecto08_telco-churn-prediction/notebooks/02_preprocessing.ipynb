```python
from src.data_preprocessing import load_data, clean_data, encode_features, scale_features

# Cargar datos
df = load_data("../data/raw/telco_churn.csv")

# Limpieza
df_clean = clean_data(df)

# Codificación de variables
df_encoded = encode_features(df_clean)

# Escalado de variables numéricas
numeric_cols = df_encoded.select_dtypes(include=['int64', 'float64']).columns.tolist()
numeric_cols.remove('Churn')

df_scaled = scale_features(df_encoded, numeric_cols)

# Guardar dataset procesado (opcional)
df_scaled.to_csv("../data/processed/telco_churn_processed.csv", index=False)
```

```python
# notebooks/02_preprocessing.ipynb
import os
import sys
import pandas as pd

sys.path.append(os.path.abspath(os.path.join(os.path.dirname("__file__"), '..')))
from src.preprocessing import load_data, clean_data, encode_categoricals, scale_features

# Cargar raw data
df = load_data("../data/raw/bank-additional.csv")

# Limpieza
df_clean = clean_data(df, drop_cols=['duration'])  # duration sesga target

# Codificación de categóricas
df_encoded = encode_categoricals(df_clean)

# Escalado
df_scaled = scale_features(df_encoded)

# Guardar processed
os.makedirs("../data/processed", exist_ok=True)
df_scaled.to_csv("../data/processed/bank_processed.csv", index=False)

print("Preprocesamiento completado. Datos guardados en data/processed")
```
